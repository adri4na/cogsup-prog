## Loading the dataset

```python
def read_examples(infile):
    """ Reads a .examples file and returns
    bows = list of BOW vectors,
                   each represented as a python dictionary,
                   for features with non-null values only
                   key = word feature
                   value = number of occurrences of the word divided by total nb of occurrences in the document
    gold_classes = list of corresponding gold classes
    """

    stream = open(infile)
    bows = []
    gold_classes = []
    bow = None
    while 1:
        line = stream.readline()
        if not line:
            break
        line = line[0:-1]
        if line.startswith("EXAMPLE_NB"):
            if bow != None:
                bows.append(bow)
            bow = {}
            cols = line.split('\t')
            gold_class = cols[3]
            gold_classes.append(gold_class)
        elif line:
            (wordform, val) = line.split('\t')
            bow[wordform] = float(val)

    if bow != None:
        bows.append(bow)
    return bows, gold_classes
```
